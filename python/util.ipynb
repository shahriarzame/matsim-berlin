```python
import pandas as pd

# ---------------- CONFIG ----------------
CSV1 = "/home/shahriar-iqbal-zame/IdeaProjects/matsim-berlin/output/newRun14/ITERS/it.0/berlin-v6.4.0.legs.csv.gz"   # replace with actual path
CSV2 = "/home/shahriar-iqbal-zame/IdeaProjects/matsim-berlin/output/newRun14/ITERS/it.5/berlin-v6.4.5.legs.csv.gz" # replace with actual path
OUT  = "/home/shahriar-iqbal-zame/IdeaProjects/matsim-berlin/output/newRun14/ITERS/output_changed_trips.csv"

# ---------------- LOAD ------------------
df1 = pd.read_csv(CSV1, sep=";")
df2 = pd.read_csv(CSV2, sep=";")

# ---------------- CHECK CHANGES ---------
# get only trip_id and mode from both
modes1 = df1[["trip_id", "mode"]].drop_duplicates()
modes2 = df2[["trip_id", "mode"]].drop_duplicates()

# merge to align same trip_id across both files
merged = pd.merge(modes1, modes2, on="trip_id", suffixes=("_csv1", "_csv2"))

# find changes from car → micro_car
changed = merged[(merged["mode_csv1"] == "micro_car") & (merged["mode_csv2"] == "car")]

# ---------------- SAVE ------------------
changed.to_csv(OUT, index=False, sep=";")
print(f"Saved {len(changed)} changed trip_ids to {OUT}")
```
